## Engagement score without loops.

### Engagement Score

- Views(0.5) -  We are using views because it tells us what is the reach and how much the video is popularity,  So we take Engagement score of 50% based on View count

- Likes(0.3) - We are using likes because it tells how much people like the video it determines how much people like the video and get interested to share and watch the video,  So we are considering with 30% percent of score based on likes.

- Comments(0.2) - We are using Comment count because it tells us people who are watching the video get interested to give comments on this video and discuss the videos with others to get the video trending.  So we are considering the 20% percent of Engagement score based on comment score.



```text
 Engagement Score= (view_count * 0.5) + (likes_count * 0.3) + (comment_count * 0.2)

````python
def Normalized_score(array : np.array) -> np.array:
  if array.size ==0:
    print('Array is Empty')
    return np.array([],dtype=float)
  normalized_score=np.zeros_like(array,dtype=float)
  denominator=array.max()-array.min()
  if denominator !=0 or denominator != 'nan':
    np.divide(array-array.min(),denominator,out=normalized_score)

  return normalized_score


def Engagement_score(views_norm_score: np.array,likes_norm_score : np.array, comment_norm_score: np.array):
  return (0.5 * views_norm_score) + (0.3 * likes_norm_score) + (0.2 * comment_norm_score)
````

````python
#Like Ratio and Comment rate in US
like_ratio_us=np.divide(likes_us_array,(likes_us_array + dislikes_us_array),where=(likes_us_array+dislikes_us_array)!=0)
print(f'Like Ratio in US: {like_ratio_us}')
comment_rate_us=np.where(views_us_array == 0,0,comment_count_us_array/views_us_array)
print(f'Comment Rate in US: {comment_rate_us}')

print()


#Like Ratio and Comment rate in IN
like_ratio_in=np.divide(likes_in_array,(likes_in_array + dislikes_in_array),where=(likes_in_array+dislikes_in_array)!=0)
print(f'Like Ratio in IN: {like_ratio_in}')
comment_rate_in=np.where(views_in_array == 0,0,comment_count_in_array/views_in_array)
print(f'Comment Rate in IN: {comment_rate_in}')


print()


#Like Ratio and Comment rate in GB
like_ratio_gb=np.divide(likes_gb_array,(likes_gb_array + dislikes_gb_array),where=(likes_gb_array+dislikes_gb_array)!=0)
print(f'Like Ratio in GB: {like_ratio_gb}')
comment_rate_gb=np.where(views_gb_array == 0,0,comment_count_gb_array/views_gb_array)
print(f'Comment Rate in GB: {comment_rate_gb}')




#Min-Max Normalization
#US
like_norm_us=Normalized_score(likes_us_array)
views_norm_us=Normalized_score(views_us_array)
comment_norm_us=Normalized_score(comment_count_us_array)

#IN
like_norm_in=Normalized_score(likes_in_array)
views_norm_in=Normalized_score(views_in_array)
comment_norm_in=Normalized_score(comment_count_in_array)

#GB
likes_norm_gb=Normalized_score(likes_gb_array)
views_norm_gb=Normalized_score(views_gb_array)
comment_norm_gb=Normalized_score(comment_count_gb_array)




#Engagement Score
engagement_score_us=Engagement_score(views_norm_us,like_norm_us,comment_norm_us)
engagement_score_in=Engagement_score(views_norm_in,like_norm_in,comment_norm_in)
engagement_score_gb=Engagement_score(views_norm_gb,likes_norm_gb,comment_norm_gb)



print(f'Engagement Score in US: {engagement_score_us}')
print(f'Engagement Score in IN: {engagement_score_in}')
print(f'Engagement Score in GB: {engagement_score_gb}')
````

### Log-Scaled View Count

- Raw views are hard to plot because the some videos get high trending and get more number of videos and some of the videos get low popularity, because of this we get highly skewed points and it is difficult to understand and train the dataset because of more videos get skewed.

- If we use the Log-Scaled View count means we will overcome the issue of skewness and make all the view count able to plot and train the model.

````python
log_view_count_us=np.log1p(views_us_array)
log_view_count_in=np.log1p(views_in_array)
log_view_count_gb=np.log1p(views_gb_array)


print(f'Log-Scaled View Count in US: {log_view_count_us}')
print(f'Log-Scaled View Count in IN: {log_view_count_in}')
print(f'Log-Scaled View Count in GB: {log_view_count_gb}')
````

## Missing values and duplicates

**Missing Values**
- We have mising values in likes, comment_count, engagement score and description columns
- Missing count
  - likes - 12 rows  (~0.02%)
  - comment_count - 36 rows (~0.07$)
  - description - 1624 rows (~3.5%)
  - Engagement score - 48 rows (~0.1%)
- **Strategy**:
  -  For `likes` and `comment_count` we can place a value of `0` because we don't know the actual count, and if we leave the null values as they are, it will create skewness and errors in statistics (mean, median, and mode). By filling with zero, we can avoid these problems.
  - For `description` we can use   `Unknown` as a placeholder, leaving as a null it will cause error in later part such as text analysis etc., Using `Unknown` it will shows as the description is missing in the dataset.
  - for `engagement score` we can recompute and fill with the values after filling the likes and comment_count null values.

**Duplicate Rows**
- Duplicate rows: More than one record with the same `video_id`, `trending_date`, and `country` are considered duplicate rows.
- In this combined dataset, 230 duplicate rows were found. We can remove these rows using the `.drop_duplicates()` function.


````python
#Checking Missing Columns and filling with appropriate values
copy_combined_df.isna().sum()
numerical_null_col=['likes','comment_count']
categorical_null_col=['description']
#for Likes, Engagement Score and comment_count we fill with 0
copy_combined_df[numerical_null_col]=copy_combined_df[numerical_null_col].fillna(0)

#for Description we fill with Unknown
copy_combined_df[categorical_null_col]=copy_combined_df[categorical_null_col].fillna('Unknown')

#Recomputing the engagement score after the null values removed
copy_combined_df['engagement_score']=Engagement_score(copy_combined_df['views'],copy_combined_df['likes'],copy_combined_df['comment_count'])
print(copy_combined_df.isna().sum())


#Checking duplicate rows and drop those rows

copy_combined_df=copy_combined_df.drop_duplicates()
print(copy_combined_df.duplicated().sum())






````

## Joining category names

I used the inner method because it merges rows only when they have matching values, whereas the left method keeps all rows from the left DataFrame even when they do not have a match.

````python
copy_combined_df=pd.merge(copy_combined_df,cat_df[['category_id','category_name']],how='inner',on='category_id')
print(copy_combined_df.shape[0])
copy_combined_df.head()
````

## Aggregation

````python
#Calculate Average views, Average likes and Average Engagement Score by category and country
average_views=copy_combined_df.groupby(['country','category_name'])['views'].agg('mean')
average_likes=copy_combined_df.groupby(['country','category_name'])['likes'].agg('mean')
average_enagement_score=copy_combined_df.groupby(['country','category_name'])['engagement_score'].agg('mean')

print("Average views:")
print(average_views)
print()
print("Average likes:")
print(average_likes)
print()
print("Average Engagement Score:")
print(average_enagement_score)



#Category Distribution Count and Percentage
category_count=copy_combined_df['category_name'].value_counts()
category_percent=copy_combined_df['category_name'].value_counts(normalize=True) *  100
print()
print("Category Count:")
print(category_count)
print()
print("Category Percentage:")
print(category_percent)


#publishing day of the week or hour tends to produce the highest-viewed videos
copy_combined_df['publish_time'] = pd.to_datetime(copy_combined_df['publish_time']).dt.tz_localize(None)
copy_combined_df['publish_day']=copy_combined_df['publish_time'].dt.day_name()
copy_combined_df['publish_hour']=copy_combined_df['publish_time'].dt.hour

views_in_days=copy_combined_df.groupby('publish_day')['views'].agg('sum').sort_values(ascending=False)
views_in_hours=copy_combined_df.groupby('publish_hour')['views'].agg('sum').sort_values(ascending=False)

print()
print(f"Views in days: {views_in_days}")
print()
print()
print("Views in hours:")
print(views_in_hours)


#trending lifespan


copy_combined_df['trending_date'] = pd.to_datetime(copy_combined_df['trending_date'], format='%y.%d.%m').dt.tz_localize(None)
copy_combined_df['days_to_trend'] = (copy_combined_df['trending_date'] - copy_combined_df['publish_time']).dt.days
print(copy_combined_df['days_to_trend'].describe())

#Using groupby() we can filter the each video takes how much days to get trending in each country
trending_lifespan=copy_combined_df.groupby(['country','video_id'])['trending_date'].nunique().reset_index()
trending_lifespan.rename(columns={'trending_date': 'trending_lifespan'},inplace=True)


#Filtering out the top 10 long lasting videos
top10_lifespan=trending_lifespan.sort_values('trending_lifespan',ascending=False)
print(top10_lifespan.head(10))
print()
#Calculating average lifespan per category.
lifespan_with_category = pd.merge(trending_lifespan, copy_combined_df[['video_id','category_name']].drop_duplicates(),on='video_id',how='left')
trending_lifespan_category=lifespan_with_category.groupby('category_name')['trending_lifespan'].nunique().mean()
print(f'The Average lifespan per category: {trending_lifespan_category}')
````
